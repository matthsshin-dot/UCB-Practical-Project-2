### Data Preparation

After our initial exploration and fine-tuning of the business understanding, it is time to construct our final dataset prior to modeling.  Here, we want to make sure to handle any integrity issues and cleaning, the engineering of new features, any transformations that we believe should happen (scaling, logarithms, normalization, etc.), and general preparation for modeling with `sklearn`. 

```python
'''
Doing some research on the problem, it is clear that there are some columns that will not affect the price.
Specifically, this includes the identifier such as id and VIN. These should be dropped. It is also prudent to drop
certain non-numeric features that are missing a lot of values, such as type and paint_color, or have too many unique values,
such as model.
'''
```

```python
df=pd.read_csv('data/vehicles.csv')
df.describe()
```

```python
#Remove any cars that are priced at less than $1000. Many of these prices do not seem to be realistic.
#Seems like some cars are basically being given away, and the price is not representative of their true properties
# Generally, the client will not be selling cars for less than $1000, so removing these outliers will allow us to get more accurate predictions for the
#rest of the data
df=pd.read_csv('data/vehicles.csv')
df=df[df['price']>=1000]

#Also going to remove cars that are over $300,000 to avoid extremely high-value vehicles skewing the model.
df=df[df['price']<300000]

#Remove columns for the VIN and ID as these do not have any bearing on the actual price
#Remove columns for paint_color and size as there are many rows missing this data
# Remove the 'models' column cause there are too many to accurately model
#Remove region as there is already a more general geographic identifier; geographic differences will be based on state if used
df_removed=df.drop(['VIN','id','model','region','paint_color','size','title_status'],axis=1)

#Remove any rows that are missing values for the:
#Condition, year, cylinders, and odometer because these are important determinants of price
df=df_removed
df=df.dropna(subset=['condition','year','odometer','cylinders','manufacturer'])
#Convert 'Cylinders' to Numerical Value, drop the rows with 'other' as the cylinder value
df['cylinders']=df['cylinders'].str.replace(" cylinders","")
df=df.query("cylinders!='other'")
df['cylinders']=pd.to_numeric(df['cylinders'])

#Convert the ordinal 'condition' data to a numerical scale
df['condition']=df['condition'].str.replace("salvage","0")
df['condition']=df['condition'].str.replace("fair","1")
df['condition']=df['condition'].str.replace("good","2")
df['condition']=df['condition'].str.replace("excellent","3")
df['condition']=df['condition'].str.replace("like new","4")
df['condition']=df['condition'].str.replace("new","5")
df['condition']=pd.to_numeric(df['condition'])

#Remove Tranmission types that are "other"
df=df.dropna(subset=['transmission'])
df=df.query("transmission!='other'")

#Remove types that are "other"
df=df.dropna(subset=['type'])
df=df.query("type != 'other'")

#Remove NaN for drive and remove 'other' values
df=df.dropna(subset=['drive'])
df=df.query("drive!='other'")

#Would like to convert the year value to an age value. Feel like the model will interpret this value more accurately.
df['year']=2026-df['year']
df=df.rename(columns={'year':'age'})

#Scale the numeric values
numeric=['age','cylinders','odometer','condition']
categorical=['manufacturer','fuel','transmission','drive','type','state']
print(df.isnull().sum())
```

```python
#These are the base features we will use for crafting the model
df
```

```python
#One-Hot Encode the categorical features

#State
state_dummies=pd.get_dummies(df['state'],dtype='int')
df=pd.concat([df,state_dummies],axis=1)
df=df.drop("state",axis=1)

#Manufacturer
manufacturer_dummies=pd.get_dummies(df['manufacturer'],dtype='int')
df=pd.concat([df,manufacturer_dummies],axis=1)
df=df.drop("manufacturer",axis=1)

#Transmission
transmission_dummies=pd.get_dummies(df['transmission'],dtype='int')
df=pd.concat([df,transmission_dummies],axis=1)
df=df.drop("transmission",axis=1)

#Fuel
fuel_dummies=pd.get_dummies(df['fuel'],dtype='int')
df=pd.concat([df,fuel_dummies],axis=1)
df=df.drop("fuel",axis=1)

#Type
type_dummies=pd.get_dummies(df['type'],dtype='int')
df=pd.concat([df,type_dummies],axis=1)
df=df.drop("type",axis=1)

#Drive
drive_dummies=pd.get_dummies(df['drive'],dtype='int')
df=pd.concat([df,drive_dummies],axis=1)
df=df.drop("drive",axis=1)
```

```python
#ALso noticed that some odometer readings are extremely high, want to cutoff the milage permitted for cars to not skew the data.
df=df[df['odometer']<300000]
```

```python
df
```

```python
# Create training and test sets
x = df.drop('price', axis=1)
y = df['price']
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

# A category for the manufacturer occurs ONLY in the test set.
variable_cols = x_train.columns[x_train.nunique(dropna=False) > 1]
removed_cols = x_train.columns.difference(variable_cols)
x_train = x_train[variable_cols].copy()
x_test = x_test[variable_cols].copy()
```

```python
#Perform Scaling on numeric features
scaler=StandardScaler()
numeric=['age','cylinders','odometer','condition']
x_train_scaled=x_train.copy()
x_test_scaled=x_test.copy()
x_train_scaled[numeric]=scaler.fit_transform(x_train_scaled[numeric])
x_test_scaled[numeric]=scaler.transform(x_test_scaled[numeric])
```

### Modeling

With your (almost?) final dataset in hand, it is now time to build some models.  Here, you should build a number of different regression models with the price as the target.  In building your models, you should explore different parameters and be sure to cross-validate your findings.

```python
#Model 1: Simple Linear Regression

#Make and fit linear regression model
m1=linear_model.LinearRegression(fit_intercept=True)
m1.fit(x_train,y_train)
#Get predictions of Model 1
m1_preds=m1.predict(x_test)
#Evaluate Model, choosing to use RMSE so that we still get our analysis in terms of dollars of error for ease of analysis but keep the benefits of MSE
RMSE_m1=np.sqrt(mean_squared_error(y_test,m1_preds))
print("Linear Model RMSE: "+ str(RMSE_m1))
#End up with an RMSE of around $8.1k. This is not completely terrible but definitely not ideal.
```

```python
# Model 2: Polynomial features with ordinary Linear Regression
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

numeric = ['age', 'condition', 'cylinders','odometer']

numeric_transformer = Pipeline([
    ('scaler', StandardScaler()),
    ('poly', PolynomialFeatures(degree=2, include_bias=False)),
    ('poly_scaler', StandardScaler())
])
preprocessor = ColumnTransformer([
    ('numeric_poly', numeric_transformer, numeric)
], remainder='passthrough')

m2 = Pipeline([
    ('preprocessor', preprocessor),
    ('regression', LinearRegression())
])
m2.fit(x_train, y_train)
m2_preds = m2.predict(x_test)
RMSE_m2 = np.sqrt(mean_squared_error(y_test, m2_preds))

print('Polynomial Regression RMSE: '+str(RMSE_m2))
```

```python
#Running a polynomial regression results in an RMSE of around $6.9k, which is notably better than the linear model
```

```python
# Model 3: Ridge Regression with Sequential Feature Selection

from sklearn.linear_model import Ridge, RidgeCV
from sklearn.feature_selection import (
    SelectKBest, f_regression, SequentialFeatureSelector
)
from sklearn.metrics import mean_squared_error
import numpy as np

sample_idx = x_train_scaled.sample(
    n=min(3000, len(x_train_scaled)),
    random_state=42
).index

X_sample = x_train_scaled.loc[sample_idx]
y_sample = y_train.loc[sample_idx]

#Identify 15 features
initial_selector = SelectKBest(
    score_func=f_regression,
    k=15
)

initial_selector.fit(X_sample, y_sample)

top_features = X_sample.columns[
    initial_selector.get_support()
].tolist()

#Run SFS on the 15 features
sfs = SequentialFeatureSelector(
    Ridge(alpha=100, solver='lsqr'),
    n_features_to_select=6,
    direction='forward',
    scoring='neg_root_mean_squared_error',
    cv=2,
    n_jobs=1
)

sfs.fit(X_sample[top_features], y_sample)

#Get selected features
selected_features = [
    f for f, selected in zip(
        top_features, sfs.get_support()
    ) if selected
]

print("Selected Features:")
print(selected_features)

#Fit Ridge on the full training dataset
m3 = RidgeCV(
    alphas=[0.1, 1, 10, 100, 1000],
    cv=3,
    scoring='neg_root_mean_squared_error'
)

m3.fit(
    x_train_scaled[selected_features],
    y_train
)

# Step 6: Evaluate on testing data
m3_preds = m3.predict(
    x_test_scaled[selected_features]
)

RMSE_m3 = np.sqrt(mean_squared_error(y_test, m3_preds))

print("Ridge RMSE: "+str(RMSE_m3))
```

```python
# Model 4: Lasso Regression

from sklearn.linear_model import LassoCV

m4 = LassoCV(
    alphas=[1, 10, 50, 100, 250, 500, 1000],
    cv=3,
    max_iter=10000,
    n_jobs=-1,
    random_state=42
)

m4.fit(x_train_scaled, y_train)

m4_preds = m4.predict(x_test_scaled)

RMSE_m4 = np.sqrt(mean_squared_error(y_test, m4_preds))

print("Lasso RMSE: "+str(RMSE_m4))
```
